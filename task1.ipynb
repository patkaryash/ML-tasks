```python
import pandas as pd
import numpy as np

# Load dataset directly from URL
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)

# Display first five records
print(df.head())

# Handle missing Age values using median
median_age = df["age"].median()
df["age"] = df["age"].fillna(median_age)

print("Missing Age values:", df["age"].isnull().sum())

# Extract surviving passengers
survivors = df[df["survived"] == 1]

# Calculate average and maximum fare using NumPy
average_fare = np.mean(survivors["fare"].dropna().to_numpy())
maximum_fare = np.max(survivors["fare"].dropna().to_numpy())

print("Total survivors:", len(survivors))
print("Average Fare:", round(average_fare, 2))
print("Maximum Fare:", round(maximum_fare, 2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar plot: survival count
sns.countplot(data=df, x="survived")
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()

# Scatter plot: Age vs Fare
sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived"
)
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```
